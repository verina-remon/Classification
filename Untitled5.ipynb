```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

df2 = pd.read_csv("Churn_Modelling.csv")

df2.drop(["RowNumber", "CustomerId", "Surname"], axis=1, inplace=True)

le = LabelEncoder()

for col in df2.select_dtypes(include='object').columns:
    df2[col] = le.fit_transform(df2[col])

X2 = df2.drop("Exited", axis=1)
y2 = df2["Exited"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)

model2 = Sequential()
model2.add(Dense(32, activation='relu', input_shape=(X2.shape[1],)))
model2.add(Dense(16, activation='relu'))
model2.add(Dense(1, activation='sigmoid'))

model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model2.fit(X2_train, y2_train, epochs=20)

y2_pred = model2.predict(X2_test)
y2_pred = (y2_pred > 0.5).astype(int)
```
