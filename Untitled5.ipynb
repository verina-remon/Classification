```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

df2 = pd.read_csv("Churn_Modelling.csv")

df2.drop(["RowNumber", "CustomerId", "Surname"], axis=1, inplace=True)

le = LabelEncoder()

for col in df2.select_dtypes(include='object').columns:
    df2[col] = le.fit_transform(df2[col])

X2 = df2.drop("Exited", axis=1)
y2 = df2["Exited"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X2_train = scaler.fit_transform(X2_train)
X2_test = scaler.transform(X2_test)

model2 = Sequential([
    Input(shape=(X2_train.shape[1],)),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model2.fit(X2_train, y2_train, epochs=20, batch_size=32, validation_split=0.2)

y2_pred = model2.predict(X2_test)
y2_pred = (y2_pred > 0.5).astype(int)
```
